```python
import scipy.ndimage as ndi
import numpy as np

def get_prompts_from_mask(mask_np):
    binary = (mask_np > 127).astype(np.uint8)
    H, W = mask_np.shape
    cy, cx = ndi.center_of_mass(binary)
    rows = np.any(binary, axis=1)
    cols = np.any(binary, axis=0)
    y_min, y_max = np.where(rows)[0][[0, -1]]
    x_min, x_max = np.where(cols)[0][[0, -1]]
    bbox_pixels = [int(x_min), int(y_min), int(x_max), int(y_max)]
    bbox_normalized = [((x_min+x_max)/2)/W, ((y_min+y_max)/2)/H,
                       (x_max-x_min)/W, (y_max-y_min)/H]
    point_as_box = [int(cx)/W, int(cy)/H, 0.01, 0.01]
    return point_as_box, bbox_normalized, bbox_pixels, "skin lesion"

def dice_score(pred, gt):
    pred, gt = pred.astype(bool), gt.astype(bool)
    intersection = (pred & gt).sum()
    if pred.sum() + gt.sum() == 0: return 1.0
    return 2 * intersection / (pred.sum() + gt.sum())

def iou_score(pred, gt):
    pred, gt = pred.astype(bool), gt.astype(bool)
    intersection = (pred & gt).sum()
    union = (pred | gt).sum()
    return intersection / union if union > 0 else 1.0

print("✅ Helper functions ready")
```

```python
from PIL import Image
import numpy as np
import cv2

TARGET_SIZE = 1024  # SAM3 expected input size

def load_and_resize(img_path, gt_mask_np):
    """Resize image and mask to 1024x1024 for SAM3."""
    # Resize image
    image = Image.open(img_path).convert("RGB")
    image = image.resize((TARGET_SIZE, TARGET_SIZE), Image.BILINEAR)

    # Resize mask
    gt_mask_resized = cv2.resize(
        gt_mask_np,
        (TARGET_SIZE, TARGET_SIZE),
        interpolation=cv2.INTER_NEAREST  # nearest neighbor keeps binary values
    )

    return image, gt_mask_resized

# Test it
img_path, gt_mask_np = dataset[0]
image, gt_mask_resized = load_and_resize(img_path, gt_mask_np)

print(f"Original mask shape: {gt_mask_np.shape}")
print(f"Resized image size:  {image.size}")
print(f"Resized mask shape:  {gt_mask_resized.shape}")
print(f"Mask unique vals:    {np.unique(gt_mask_resized)}")  # still [0, 255]
print(f"Coverage before:     {(gt_mask_np > 127).mean()*100:.1f}%")
print(f"Coverage after:      {(gt_mask_resized > 127).mean()*100:.1f}%")  # should be ~same
```

```python
def run_inference_all_prompts(processor, image, gt_mask_np):
    results = {}
    point_box, bbox_norm, bbox_pix, text = get_prompts_from_mask(gt_mask_np)
    H, W = gt_mask_np.shape

    def extract_best_mask(out, shape):
        masks  = out["masks"]
        scores = out["scores"]
        if masks is None or masks.numel() == 0 or len(masks) == 0:
            return np.zeros(shape, dtype=bool)
        best_idx = scores.argmax().item()
        return masks[best_idx].cpu().numpy().squeeze() > 0

    # ── 1. TEXT PROMPT ──────────────────────────────────────────────
    try:
        with torch.autocast(device_type="cuda", dtype=torch.bfloat16):
            state = processor.set_image(image)
            out = processor.set_text_prompt(state=state, prompt="skin lesion")
        results["text"] = extract_best_mask(out, (H, W))
    except Exception as e:
        print(f"  ❌ text failed: {e}")
        results["text"] = np.zeros((H, W), dtype=bool)

    # ── 2. BOUNDING BOX PROMPT ──────────────────────────────────────
    try:
        with torch.autocast(device_type="cuda", dtype=torch.bfloat16):
            state = processor.set_image(image)
            out   = processor.add_geometric_prompt(
                        box=bbox_norm, label=True, state=state)
        results["bbox"] = extract_best_mask(out, (H, W))
    except Exception as e:
        print(f"  ❌ bbox failed: {e}")
        results["bbox"] = np.zeros((H, W), dtype=bool)

    # ── 3. POINT PROMPT ─────────────────────────────────────────────
    try:
        with torch.autocast(device_type="cuda", dtype=torch.bfloat16):
            state = processor.set_image(image)
            state = processor.set_text_prompt(state=state, prompt="skin lesion")
            px, py = point_box[0], point_box[1]
            out   = processor.add_geometric_prompt(
                        box=[px, py, 0.10, 0.10], label=True, state=state)
        results["point"] = extract_best_mask(out, (H, W))
    except Exception as e:
        print(f"  ❌ point failed: {e}")
        results["point"] = np.zeros((H, W), dtype=bool)

    # ── 4. AUTOMATIC PROMPT ─────────────────────────────────────────
    try:
        with torch.autocast(device_type="cuda", dtype=torch.bfloat16):
            state = processor.set_image(image)
            out   = processor.set_text_prompt(state=state,
                                              prompt="dermoscopic skin lesion")
        results["auto"] = extract_best_mask(out, (H, W))
    except Exception as e:
        print(f"  ❌ auto failed: {e}")
        results["auto"] = np.zeros((H, W), dtype=bool)

    return results
```

```python
import torch.nn.functional as F

def dice_loss(pred_mask, gt_mask):
    """
    pred_mask: float tensor [H, W] — raw logits or probabilities
    gt_mask:   float tensor [H, W] — binary 0/1
    """
    pred  = torch.sigmoid(pred_mask).flatten()
    gt    = gt_mask.float().flatten()
    inter = (pred * gt).sum()
    return 1 - (2 * inter + 1) / (pred.sum() + gt.sum() + 1)

def bce_loss(pred_mask, gt_mask):
    return F.binary_cross_entropy_with_logits(
        pred_mask.flatten(),
        gt_mask.float().flatten()
    )

def combined_loss(pred_mask, gt_mask):
    """Dice + BCE — standard medical segmentation loss."""
    return dice_loss(pred_mask, gt_mask) + bce_loss(pred_mask, gt_mask)

print("✅ Loss functions defined")
```

```python
import torch.nn.functional as F

def forward_with_grad(model, processor, image, bbox_norm, H, W):

    # 1. Preprocess image
    img_tensor = v2.functional.to_image(image).to(processor.device)
    img_tensor = processor.transform(img_tensor).unsqueeze(0)

    # 2. Wrap the ENTIRE forward pass in autocast (Just like your inference code!)
    # We use float16 because you are on a T4 GPU which handles float16 natively
    with torch.autocast(device_type="cuda", dtype=torch.bfloat16):

        # Backbone — frozen, use no_grad
        with torch.no_grad():
            backbone_out = model.backbone.forward_image(img_tensor)
            dummy_text   = model.backbone.forward_text(
                               ["visual"], device=processor.device)
            backbone_out.update(dummy_text)

        # Geometric prompt
        boxes  = torch.tensor(bbox_norm, device=processor.device,
                              dtype=torch.float32).view(1, 1, 4)
        labels = torch.tensor([True], device=processor.device,
                              dtype=torch.bool).view(1, 1)
        geo_prompt = model._get_dummy_prompt()
        geo_prompt.append_boxes(boxes, labels)

        # Forward grounding
        outputs = model.forward_grounding(
            backbone_out     = backbone_out,
            find_input       = processor.find_stage,
            geometric_prompt = geo_prompt,
            find_target      = None,
        )

        # 3. Extract outputs and cast to float32 explicitly ONLY at the very end
        # for stable loss calculation
        out_masks  = outputs["pred_masks"].float()
        out_logits = outputs["pred_logits"].float()
        out_probs  = out_logits.sigmoid()
        presence   = outputs["presence_logit_dec"].float().sigmoid().unsqueeze(1)
        out_probs  = (out_probs * presence).squeeze(-1)

    # 4. Interpolate outside of autocast to prevent complex coordinate math issues
    if out_masks.dim() == 3:
        out_masks = out_masks.unsqueeze(1)

    out_masks = F.interpolate(out_masks, (H, W),
                              mode="bilinear", align_corners=False)

    return out_masks, out_probs
```

```python
import torch.nn.functional as F
from torchvision.transforms import v2

# 1. Load and prepare the data
img_path, gt_mask_np = dataset[0]
image, gt_mask_np    = load_and_resize(img_path, gt_mask_np)
point_box, bbox_norm, bbox_pix, text = get_prompts_from_mask(gt_mask_np)
gt_bin = torch.tensor(gt_mask_np > 127, dtype=torch.float32).to('cuda')
H, W   = gt_mask_np.shape

# 2. Force the model into eval mode to bypass the internal target-hunting crash
model.eval()

# 3. Run the forward pass! (Gradients will STILL be tracked!)
masks, scores = forward_with_grad(model, processor, image, bbox_norm, H, W)

# 4. ── THE BATCH DIMENSION FIX ──
# We must specify index [0] to look inside the first image of the batch
best_idx = scores[0].argmax().item()
logit    = masks[0, best_idx].squeeze()

print("logit requires_grad:", logit.requires_grad)
print("logit dtype:",         logit.dtype)
print("logit shape:",         logit.shape)

# 5. Calculate your external loss
loss = F.binary_cross_entropy_with_logits(
    logit.flatten(), gt_bin.flatten()
)

print("loss:",                round(loss.item(), 4))
print("loss requires_grad:", loss.requires_grad)

# 6. Clear any old gradients just in case, then run backward
model.zero_grad()
loss.backward()

print("✅ Backward pass works!")
```

```python
import gc
import numpy as np
import random
import torch.nn.functional as F
from tqdm import tqdm

# ── 1. Optimizer & Scheduler Setup ───────────────────────────────────
# We only want to update parameters that have requires_grad=True (your decoder)
optimizer = torch.optim.AdamW(
    [p for p in model.parameters() if p.requires_grad],
    lr=1e-4,
    weight_decay=0.01
)

scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(
    optimizer, T_max=10, eta_min=1e-6
)

EPOCHS     = 15
BEST_DICE  = 0.0
train_log  = []

print(f"🚀 Starting fine-tuning for {EPOCHS} epochs on {len(train_dataset)} images\n")

# ── 2. Main Training Loop ────────────────────────────────────────────
for epoch in range(EPOCHS):

    # 🚨 CRITICAL: Keep model in eval mode to prevent the 'find_target' crash!
    # Gradients are STILL tracked because we set requires_grad=True earlier.
    model.eval()

    epoch_losses = []
    random.shuffle(train_dataset)

    # Wrap the dataset in tqdm for a nice progress bar
    for img_path, gt_mask_np in tqdm(train_dataset, desc=f"Epoch {epoch+1}/{EPOCHS}"):
        try:
            # Load and resize image and ground truth
            image, gt_mask_resized = load_and_resize(img_path, gt_mask_np)
            H, W = gt_mask_resized.shape

            # Convert ground truth to binary tensor on GPU
            gt_bin = torch.tensor(
                gt_mask_resized > 127, dtype=torch.float32
            ).to('cuda')

            # Skip entirely empty masks to prevent NaN loss
            if gt_bin.sum() == 0:
                continue

            optimizer.zero_grad()

            # Get the bounding box prompt from your ground truth
            point_box, bbox_norm, bbox_pix, text = get_prompts_from_mask(gt_mask_resized)

            # ── 3. Forward Pass (Using our custom grad-tracking function!) ──
            masks_logits, scores = forward_with_grad(model, processor, image, bbox_norm, H, W)

            if masks_logits is None or masks_logits.numel() == 0:
                continue

            # ── 4. Extract Best Mask using the Batch Dimension Fix [0] ──
            best_idx = scores[0].argmax().item()
            logit    = masks_logits[0, best_idx].squeeze()  # Output is exactly [H, W]

            # Resize logit to perfectly match GT dimensions if they drifted
            if logit.shape != (H, W):
                logit = F.interpolate(
                    logit.unsqueeze(0).unsqueeze(0).float(),
                    size=(H, W), mode='bilinear', align_corners=False
                ).squeeze()

            # ── 5. Loss & Backward Pass ──
            loss = combined_loss(logit, gt_bin)
            loss.backward()

            # Gradient clipping to prevent exploding gradients
            torch.nn.utils.clip_grad_norm_(
                [p for p in model.parameters() if p.requires_grad],
                max_norm=1.0
            )

            optimizer.step()
            epoch_losses.append(loss.item())

        except Exception as e:
            # Print the error so you know if a specific corrupted image is failing
            print(f"\n⚠️ Error on {img_path}: {e}")
            continue

       # finally:
            # Keep VRAM clean so Colab doesn't crash
            #torch.cuda.empty_cache()
            #gc.collect()

    # ── 6. Epoch Summary & Scheduler ─────────────────────────────────
    scheduler.step()
    avg_loss = np.mean(epoch_losses) if epoch_losses else 0

    # ── 7. Quick Validation on 50 Test Images ────────────────────────
    val_dices = []

    for img_path, gt_mask_np in test_dataset[:200]:
        try:
            image, gt_mask_resized = load_and_resize(img_path, gt_mask_np)
            gt_bin = gt_mask_resized > 127

            # For validation, we DO NOT need gradients, so we safely use the processor
            with torch.no_grad():
                with torch.autocast(device_type="cuda", dtype=torch.bfloat16):
                    state = processor.set_image(image)
                    out   = processor.add_geometric_prompt(
                                box=get_prompts_from_mask(gt_mask_resized)[1],
                                label=True, state=state)

                masks  = out["masks"]
                scores = out["scores"]

                if masks is not None and len(masks) > 0:
                    best   = scores[0].argmax().item()
                    mask   = masks[0, best].cpu().numpy().squeeze() > 0
                    val_dices.append(dice_score(mask, gt_bin))

        except:
            continue
        #finally:
           # torch.cuda.empty_cache()

    avg_val_dice = np.mean(val_dices) if val_dices else 0

    # Print out how the epoch did!
    print(f"\n✅ Epoch {epoch+1:2d} | Train Loss: {avg_loss:.4f} | "
          f"Val Dice (bbox): {avg_val_dice:.4f} | "
          f"LR: {scheduler.get_last_lr()[0]:.2e}")

    train_log.append({
        "epoch": epoch + 1,
        "loss":  round(avg_loss, 4),
        "val_dice": round(avg_val_dice, 4)
    })

    # ── 8. Save Best Checkpoint to Google Drive ──────────────────────
    if avg_val_dice > BEST_DICE:
        BEST_DICE = avg_val_dice

        # Save everything needed to resume training later
        torch.save({
            "epoch":       epoch + 1,
            "model_state": model.state_dict(),
            "optimizer":   optimizer.state_dict(),
            "best_dice":   BEST_DICE
        }, '/content/drive/MyDrive/sam3_finetuned_best.pt')

        print(f"  💾 New best model saved to Drive! Dice: {BEST_DICE:.4f}")

print(f"\n🎉 Fine-tuning complete! Best Val Dice: {BEST_DICE:.4f}")
```

```python
import matplotlib.pyplot as plt
import numpy as np
import cv2
import random

def visualize_10_images(dataset, processor, n=10):

    fig, axes = plt.subplots(n, 6, figsize=(24, 4 * n))

    col_titles = ["Original", "GT Mask",
                  "Text", "BBox", "Point", "Auto"]
    for col, title in enumerate(col_titles):
        axes[0, col].set_title(title, fontsize=12,
                               fontweight='bold', pad=10)

    for row in range(n):
        img_path, gt_mask_np = dataset[row]
        image, gt_mask_resized = load_and_resize(img_path, gt_mask_np)
        gt_bin = gt_mask_resized > 127
        H, W   = gt_mask_resized.shape

        # ── Col 0: Original ──────────────────────────────────────────
        axes[row, 0].imshow(image)
        axes[row, 0].set_ylabel(img_path.stem[:12],
                                fontsize=7, rotation=0,
                                labelpad=75, va='center')
        axes[row, 0].axis('off')

        # ── Col 1: GT Mask ───────────────────────────────────────────
        axes[row, 1].imshow(image)
        gt_overlay = np.ma.masked_where(~gt_bin, gt_bin)
        axes[row, 1].imshow(gt_overlay, cmap='cool', alpha=0.6)
        axes[row, 1].set_xlabel(
            f"Coverage: {gt_bin.mean()*100:.1f}%", fontsize=8)
        axes[row, 1].axis('off')

        # ── Run all 4 prompts ────────────────────────────────────────
        preds = run_inference_all_prompts(processor, image, gt_mask_resized)

        prompt_cols = {
            "text":  (2, 'spring'),
            "bbox":  (3, 'autumn'),
            "point": (4, 'winter'),
            "auto":  (5, 'plasma')
        }

        for prompt_type, (col, cmap) in prompt_cols.items():
            ax   = axes[row, col]
            mask = preds[prompt_type]

            if mask.shape != (H, W):
                mask = cv2.resize(
                    mask.astype(np.uint8), (W, H)).astype(bool)

            d   = dice_score(mask, gt_bin)
            iou = iou_score(mask, gt_bin)

            ax.imshow(image)
            if mask.sum() > 0:
                overlay = np.ma.masked_where(~mask, mask)
                ax.imshow(overlay, cmap=cmap, alpha=0.55)

            # GT contour in white
            contours, _ = cv2.findContours(
                gt_bin.astype(np.uint8),
                cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
            img_np = np.array(image)
            cv2.drawContours(img_np, contours, -1, (255,255,255), 2)

            color = ('green' if d >= 0.7 else
                     'orange' if d >= 0.4 else 'red')
            ax.set_xlabel(f"Dice: {d:.3f}\nIoU: {iou:.3f}",
                         fontsize=8, color=color)
            ax.axis('off')

        torch.cuda.empty_cache()

    plt.suptitle(
        "SAM3 Fine-tuned Model — All 4 Prompts on 10 ISIC Images\n"
        "White contour = GT boundary | Color = predicted mask",
        fontsize=13, fontweight='bold', y=1.005)
    plt.tight_layout()
    plt.savefig('/content/drive/MyDrive/sam3_10image_visualization.png',
                bbox_inches='tight', dpi=120)
    plt.show()
    print("✅ Saved to Drive!")

# Run on first 10 images
# Pick 10 random indices from dataset
random_indices = random.sample(range(len(dataset)), 10)
random_subset  = [dataset[i] for i in random_indices]

# Run visualization on random subset
visualize_10_images(random_subset, processor, n=10)
```
